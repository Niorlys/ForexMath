# Stochastic processes

In this note, we discuss the mathematical tools needed to model continous-time markets, and making use of the stochastic engines that serve us as source of randomness in continuous-time. 
 
>**Definition (Stochastic process at continuous time)** A stochastic process is a collection of random variables indexed by a continuous set $T$, i.e., $\{X_t: t\in T\}$.

Here, as before when dealing with the discrete case, we get a trajectory for a fixed $\omega\in\Omega$ through the evaluation of the function $\mathbb T_\omega(t):t\to X_t(\omega)$ from $T$ to $\R$. Also, we must consider the concept of distribution of a stochastic process. 
 
>**Definition (Finite-dimensional distribution)** Let $\vec{t}=(t_1,t_2,\ldots,t_n)\in T^n$ be a vector of times, then the joint distribution of the vector of random variables $(X_{t_1}, X_{t_2}, \ldots, X_{t_n})$ is called the finite dimensional distribution, and is denoted as $F_{t_1,t_2,\dots,t_n}$. The collection $\{F_{t_1,t_2,\dots,t_n}\}$ as $\vec{t}$ ranges over vectors in $T^n$, for all $n\in\N$, is called the collection of finite dimensional distributions(f.d.ds).

>**Definition (Strong stationarity)** A stochastic process is said to be strongly stationary if the finite-dimensional distributions are invariant under time shifts, i.e., for all $t_1,t_2,\ldots,t_n\in T$ and $h\in T$,
>$$
>F_{t_1+h,t_2+h,\ldots,t_n+h}=F_{t_1,t_2,\ldots,t_n}.
>$$

>**Definition (Weak stationarity)** A stochastic process is said to be weakly stationary if its mean and covariance function are invariant under time shifts, i.e., for all $t_1,t_2,\ldots,t_n\in T$ and $h\in T$,
>$$
>\mathbb{E}[X_{t_1}]=\mathbb{E}[X_{t_1+h}],\quad \text{and}\quad Cov(X_{t_1},X_{t_2})= Cov(X_{t_1+h},X_{t_2+h}).
>$$


A natural question to ask is: given a family of f.d.ds, can we find a stochastic process that has those f.d.ds? The answer is yes, and we can do this by using the Kolmogorov's consistency theorem.

>**Theorem (Kolmogorov's consistency theorem)** Let $\mathbb{F}=\{F_{t_1,t_2,\dots,t_n}\}$ be a family of f.d.ds, $n\in\N$ and $T$ a non-empty set. Then there exists a probability space $(\Omega, \mathcal{F}, \mathbb{P})$ and a stochastic process $\{X_t: t\in T\}$  on that space such that the f.d.ds of $\{X_t\}$ are exactly $\mathbb{F}$.

**Proof** User can find the proof in Krishna's book "Measure Theory and Probability Theory" in Chapter 6, Section 3. $\blacksquare$

To construct the stochastic process, recall that we can identify any outcome $\omega$ from the sample space with a function $f_{\omega}:T\to\R$ defined as $f_{\omega}(t)=X_t(\omega)$, this is, when we had the process, we could define the $f_{\omega}$ as the trajectory of the process. Now, we must go backwards, in other words, we have the index set $T$, so we can define the stochastic process based on the trajectories. Therefore, it is natural to take as sample space for the process the set of all functions $\omega:T\to\R$, which we'll denote as $\R^T$, and the analytical expression for the (canonical) stochastic process is given by
$$
X_t(\omega)=\omega(t).
$$

>**Definition (Canonical stochastic process)** The stochastic process $\{X_t: t\in T\}$ defined by $X_t(\omega)=\omega(t)$ for all $t\in T$ is called the canonical stochastic process.

Now, we need to define $\mathcal{F}$. To do this, it is strightforward that for a Borel set $B\in\mathcal{B}(\R^k)$,
$$
(X_{t_1}(\omega),X_{t_2}(\omega),\ldots,X_{t_k}(\omega))\in B\iff (\omega(t_1),\omega(t_2),\ldots,\omega(t_k))\in B.
$$

The right hand side of the above equivalence is quite important to ensure measurability for the canonical stochastic process, and leads us to the definition of **cylinders**, where we have a directrix consisting in the Borel set $B$ and a generatrix consisting in the vector of indexes or times $(t_1,t_2,\ldots,t_k)$. 
 
>**Definition (Cylinder)** For the set of times $T$, $B\in\mathcal{B}(\R^k)$ and $A=\{t_1,t_2,\ldots,t_k\}\subset T$. The set $C(B,A)\subset\R^T$ is called finite dimensional cylinder, where
>$$
>C(B,A)=\{\omega\in\R^T: (\omega(t_1),\omega(t_2),\ldots,\omega(t_k))\in B\}.
>$$
> The collection of all finite dimensional cylinders over a finite $A\subset T$ and $B\in\mathcal{B}(\R^{|A|})$ is denoted as $\mathcal{C}_T$.


Then, we can safely take $\mathcal{F}=\sigma(\mathcal{C}_T)$. For a particular cylinder $C(B,A)$, when $B=(-\infty,x_1]\times(-\infty,x_2]\times\ldots\times(-\infty,x_k]$, we can assign probabilities to the cylinders according to the f.d.ds as follows:
>$$
>\mathbb{P}(C(B,A))=\mathbb{P}(X_{t_1}\leq x_1,X_{t_2}\leq x_2,\ldots,X_{t_k}\leq x_k)=F_{t_1,t_2,\ldots,t_k}(x_1,x_2,\ldots,x_k).
>$$

Having constructed the canonical stochastic process, and the probability space it lives in, we ask ourselves: are $\omega(t)$ continuous a.s.? This question at first, doesn't make sense, since the set $F=\{\omega\in\R^T: \omega(t)\text{ is continuous}\}$ is not measureble respect to $\mathcal{F}$ mentioned above (continuity involves an infinite number of elements of $T$). However, with a proper twist, we can make sense of this question. To this end, we define the following:

>**Definition (Continuous stochastic process)** A stochastic process $\{X_t: t\in T\}$ is said to be continuous if the set $F=\{\omega\in\R^T: \omega(t)\text{ is not continuous}\}$ has probability zero, i.e., $\mathbb{P}(F)=1$.

>**Definition (Version of a stochastic process)** A version of a stochastic process $\{X_t: t\in T\}$ is a stochastic process $\{Y_t: t\in T\}$ such that $Y_t(\omega)=X_t(\omega)$ a.s. for all $t\in T$.

Note that two stochastic processes being versions of each other means that they are indistinguishable in the sense that they have the same finite-dimensional distributions. The continuity question can be answered by the following theorem: 
 
>**Theorem (Kolmogorov's continuity theorem)** Let $\{X_t: t\in T\}$ be a stochastic process such that for all $T>0$, there exist positive constants $\alpha, \beta$ and $W$ such that
>$$
>\mathbb{E}[|X_t-X_s|^{\alpha}]\leq W|t-s|^{1+\beta},
>$$
>for all $t,s\in T$ with $t\geq s$. Then, there exists a version of $\{X_t\}$ that is continuous.


Then, when we have a stochastic process whose expectation decays as specified in the theorem, we can restrict ourselves to take $\Omega$ as the space of continuous functions on $T$. 
 
To describe the flow of information in continuous-time stochastic processes, for a given time $t$, we consider the cylinders generated by the collections of times $t_1,t_2,\ldots,t_k$ with $t_1<t_2<\ldots<t_k\leq t$, and any Borel sets $B_1,B_2,\ldots,B_k\in\mathcal{B}(\R)$. Let $C_t$ be the collection of such cylinders, we take $\mathcal{F}_t=\sigma(C_t)$. Note that for $t'<t$, we have that $\mathcal{F}_t'\subseteq\mathcal{F}_{t}$. So, the collection of $\sigma$-algebras $\{\mathcal{F}_t\}$ forms a **filtration** which is generated by the canonical stochastic process. The filtration can be used to define martingales in continuous-time.

>**Definition (Martingale in continuous time)** A stochastic process $\{X_t: t\in T\}$ is said to be a martingale if 
>
>i. $X_t\in\mathcal{F}_t$ for all $t\in T$,
>
>ii. $\mathbb{E}[|X_t|]<\infty$ for all $t\in T$,
>
>iii. $\mathbb{E}[X_t|\mathcal{F}_{t'}]=X_{t'}$, $t'<t$.

It is called supermartingale(submartingale) when in iii. we replace  $=$ with $\leq$ (resp. $\geq$).


>**Definition (Characteristic function of a stochastic process)** The characteristic function of a stochastic process $\{X_t: t\in T\}$ is defined as
>$$
>\phi_X(t, s)=\mathbb{E}[e^{isX_t}]
>$$

## Markov Processes

A stochastic process is said to have the Markov property when the future depends only on the present, and not on the past. This mean that the probability of a future behaviour of the process, when the present state is known, is not altered by additional information about the past. We remark that when the present state is not known with certainty, then in general, the probability of a future behaviour of the process is altered by additional information about the past. Formally, 

>**Definition (Markov property)** A stochastic process $\{X_t: t\in [0,T]\}$ is said to have the Markov property if for all $t_1,t_2,\ldots,t_n, t\in [0,T]$ and $B_1,B_2,\ldots,B_n, B\in\mathcal{B}(\R)$,
>$$
>\mathbb{P}(X_t\in B|X_{t_{n}}\in B_{n},\ldots,X_{t_1}\in B_1)=\mathbb{P}(X_{t}\in B|X_{t_{n}}\in B_{n}).
>$$
>whenever $t_1<t_2<\ldots<t_n<t$.

Now,  if we encode the past and present information using the filtration $\mathcal{F}_t$ we defined in the previous section, we can write the Markov property as: 

>**Definition (Markov property via filtrations)** A stochastic process $\{X_t: t\in [0,T]\}$ is said to have the Markov property if for all $t'< t\leq T$ and $B\in\mathcal{B}(\R)$,
>$$
>\mathbb{P}(X_t\in B|\mathcal{F}_{t'})=\mathbb{P}(X_{t}\in B|X_{t'}).
>$$



Recall that since $X_{t'}$ is a real random variable, and $\R$ is a Polish space, we can think of the above conditional probability as a regular conditional probability. A basic concept for the study of the structure of Markov processes is the concept of **transition probability function(tpf)**.

>**Definition (Transition probability)** The transition probability function of a Markov process $\{X_t: t\in T\}$ is defined as
>$$
>P(t',x,t,B)=\mathbb{P}(X_t\in B|X_{t'}=x).
>$$
>for all $t',t\in T$, $t'<t$, $x\in\R$ and $B\in\mathcal{B}(\R)$.

Note that transition probability is just the r.c.p evaluated for $\omega\in X_{t'}^{-1}(x)$. This means that for fixed $t',t$ and $x$, the set function $P(t',x,t,\cdot)$ is a probability measure on $\R$. Another important property of the transition probability is the Chapman-Kolmogorov equation.

>**Theorem (Chapman-Kolmogorov equation)** For all $t',t,s\in T$, $t'<s<t$, and $B\in\mathcal{B}(\R)$,
>$$
>P(t',x',t,B)=\int P(s,x'',t,B)P(t',x',s,dx'').
>$$

**Proof:**  
 
Consider the nested $\sigma$-algebras $\sigma(X_{t'}) \subset \mathcal{F}_s$. The Tower Property tells us:
    $$P(t', x', t, B) = \mathbb{E}[  \mathbb{P}(X_t \in B  |  \mathcal{F}_s)  |  X_{t'} = x'  ]. \tag{1}$$

The Markov Property allows us to replace the entire history $\mathcal{F}_s$ with the present state $\sigma(X_s)$, so we have
    $$P(t', x', t, B) = \mathbb{E}[  \mathbb{P}(X_t \in B  |  X_s)  |  X_{t'} = x'  ]. \tag{2}$$

Let $\mu(\omega, d\omega')$ be the regular conditional probability on $\mathcal{G}$ given $\sigma(X_{t'})$. For a fixed $\omega$ where $X_{t'}(\omega)=x$, we apply Proposition 3.35 to the random variable $Y = \mathbb{P}(X_t \in B  |  X_s)$:
    $$
    \mathbb{E}[Y  |  \sigma(X_{t'})](\omega) = \int_{\Omega} Y(\omega')  \mu(\omega, d\omega').
    $$
    By definition, $Y(\omega') = \mathbb{P}(X_t \in B  |  X_s)(\omega') = P(s, X_s(\omega'), t, B)$. Therefore, the equation becomes:
    $$
    \mathbb{E}[  \mathbb{P}(X_t \in B  |  X_s)  |  X_{t'} = x'  ] = \int_{\Omega} P(s, X_s(\omega'), t, B)  \mu(\omega, d\omega'). \tag{3}
    $$

The measure $\mu(\omega, \cdot)$ is the conditional distribution of an outcome $\omega$ given $X_{t'} = x'$. The integral over $\Omega$ in (3) can be transformed into an integral over the state space $\mathbb{R}$ by considering the distribution of $X_s$ under this measure.
    Let $\nu(\omega, dx'')$ be the pushforward measure on $\mathbb{R}$ defined by $\nu(\omega, A) = \mu(\omega, \{ \omega' : X_s(\omega') \in A \})$ for any Borel set $A$. By definition, $\nu(\omega, dx'')$ is the regular conditional distribution of $X_s$ given $X_{t'}=x'$, which is precisely $P(t', x', s, dx'')$.
    Performing this change of variables, the integral in (3) becomes:
    $$
    \int_{\mathbb{R}} P(s, x'', t, B)  P(t', x', s, dx'').
    $$

Substituting this result back into equation (2) gives the final form of the Chapman-Kolmogorov equation:
    $$
    P(t', x', t, B) = \int_{\mathbb{R}} P(s, x'', t, B)  P(t', x', s, dx'').
    $$
    $\blacksquare$


In order to the process to be stationary, we need that the transition probability depends only on the difference between the times.
 
 >**Definition (Homogeneous Markov process)** A Markov process is said to be homogeneous if its transition probability function depends only on the length of the time interval, i.e.,
 >$$
 >P(t',x',t,B)=P(t-t',x',B),
 >$$
 >for all $t',t\in T$, $t'<t$, $x\in\R$ and $B\in\mathcal{B}(\R)$.



For a homogeneous Markov process, the Chapman-Kolmogorov equations take the form
>$$
>P(t-t',x',B)=\int P(t-s,x'',B)P(s-t',x',dx'').
>$$

>**Definition (Transition probability density function(tpdf))** We say that the t.p.f. $P(t',x,t,B)$ has a density $p(t',x,t,B)$ if
>$$
>P(t',x',t,B)=\int p(t',x',t,x)dx.
>$$

When we have t.p.d.f, the Chapman-Kolmogorov equation takes the form
>$$
>p(t', x', t, x) = \int_{\mathbb{R}} p(s, x'', t, x)  p(t', x', s, x'')dx''
>$$

and the corresponding equation for an homogeneous Markov process is
>$$
>p(t-t', x', t, x) = \int_{\mathbb{R}} p(t-s, x'', t, x)  p(s-t', x', s, x'')dx''
>$$


**Example (Wiener process)** The Wiener process the main example of Markov process with continuous paths. It is defined as a stochastic process $\{W_t: t\in [0,T]\}$ such that $W_0=0$ and have stationary t.p.d.f. given by
>$$
>p(\tau, x, x') = \frac{1}{\sqrt{2\pi \tau}} e^{-\frac{(x'-x)^2}{2\tau}}\equiv H(\tau, x'- x).
>$$

To see this, let's consider times $t < t' < t''$, let $\tau = t' - t$ and $\sigma = t'' - t'$, so $t'' - t = \tau + \sigma$. The Chapman–Kolmogorov equation requires:

$$
p(\tau + \sigma, x, x'') = \int_{-\infty}^{\infty} p(\tau, x, x') \, p(\sigma, x', x'') \, dx'.
$$

Substitute the expressions:
\begin{align*}
\frac{1}{\sqrt{2\pi(\tau + \sigma)}} \exp\left(-\frac{(x'' - x)^2}{2(\tau + \sigma)}\right) &= \int_{-\infty}^{\infty} \frac{1}{\sqrt{2\pi\tau}} \exp\left(-\frac{(x' - x)^2}{2\tau}\right) \cdot \frac{1}{\sqrt{2\pi\sigma}} \exp\left(-\frac{(x'' - x')^2}{2\sigma}\right) dx'\\
&= \frac{1}{2\pi\sqrt{\tau\sigma}} \int_{-\infty}^{\infty} \exp\left[-\frac{1}{2} \left( \frac{(x' - x)^2}{\tau} + \frac{(x'' - x')^2}{\sigma} \right) \right] dx' \equiv \text{RHS}
\end{align*}


Let  $A = \frac{1}{\tau} + \frac{1}{\sigma} = \frac{\tau + \sigma}{\tau\sigma} $, and $B = \frac{x}{\tau} + \frac{x''}{\sigma} $. Completing the square in the exponent:

$$
\frac{(x' - x)^2}{\tau} + \frac{(x'' - x')^2}{\sigma} = A \left(x' - \frac{B}{A}\right)^2 - \frac{B^2}{A} + \frac{x^2}{\tau} + \frac{x''^2}{\sigma}.
$$

The integral becomes:

$$
\int_{-\infty}^{\infty} \exp\left(-\frac{1}{2} A \left(x' - \frac{B}{A}\right)^2 \right) dx' = \sqrt{\frac{2\pi}{A}},
$$

so:


$$
\text{RHS} = \frac{1}{2\pi\sqrt{\tau\sigma}} \cdot \sqrt{\frac{2\pi}{A}} \cdot \exp\left( \frac{1}{2} \frac{B^2}{A} - \frac{1}{2} \left( \frac{x^2}{\tau} + \frac{x''^2}{\sigma} \right) \right).
$$

Note that $ \tau\sigma A = \tau + \sigma $, so:

$$
\frac{1}{2\pi\sqrt{\tau\sigma}} \cdot \sqrt{\frac{2\pi}{A}} = \frac{1}{\sqrt{2\pi(\tau + \sigma)}}.
$$

A computation shows:

$$
\frac{1}{2} \frac{B^2}{A} - \frac{1}{2} \left( \frac{x^2}{\tau} + \frac{x''^2}{\sigma} \right) = -\frac{(x'' - x)^2}{2(\tau + \sigma)},
$$

thus:

$$
\text{RHS} = \frac{1}{\sqrt{2\pi(\tau + \sigma)}} \exp\left(-\frac{(x'' - x)^2}{2(\tau + \sigma)}\right)
$$
 therefore the Chapman-Kolmogorov equation is satisfied. Kolmogorov’s continuity theorem states that if for some $ \alpha, \beta, C > 0 $:
$$
\mathbb{E}[|W_t - W_s|^\alpha] \leq C |t - s|^{1 + \beta},
$$

then the process has continuous trajectories.

For the Wiener process, $ W_t - W_s \sim \mathcal{N}(0, t - s) $. For $ \alpha = 4 $:

$$
\mathbb{E}[(W_t - W_s)^4] = 3(t - s)^2.
$$

Thus:

$$
\mathbb{E}[|W_t - W_s|^4] = 3 |t - s|^2,
$$

which satisfies the condition with $ \alpha = 4 $, $ \beta = 1 $, and $ C = 3 $. So, the Wiener process has continuous trajectories, however, we will see later that they are nowhere differentiable. Note that for a fixed $t$, $W_t\sim\mathcal{N}(0,t)$. The function $H(\tau, x'- x)$ is called the **heat kernel**. The reader can extend the Chapman-Kolmogorov equation to the case of $\R^n$, likewise the Wiener process in $\R^n$.
